```python
# This is necessary to recognize the modules
import os
import sys
from decimal import Decimal
import warnings

warnings.filterwarnings("ignore")

root_path = os.path.abspath(os.path.join(os.getcwd(), '../..'))
sys.path.append(root_path)
```

```python
from core.data_sources.clob import CLOBDataSource

# Get trading rules and candles
clob = CLOBDataSource()
```

```python
# Constants
CONNECTOR_NAME = "binance"
INTERVALS = ["1h"]
trading_pairs = ['ETH-USDT']
DAYS = 3600
```

# Download data
- Get trading rules
- Get candles for the last x days

```python
BATCH_CANDLES_REQUEST = 10
SLEEP_REQUEST = 0.65

all_candles = {
    interval: await clob.get_candles_batch_last_days(CONNECTOR_NAME, trading_pairs, interval, DAYS, BATCH_CANDLES_REQUEST,
                                                  SLEEP_REQUEST) for interval in INTERVALS
}
clob.dump_candles_cache(root_path)
```

```python
from core.data_structures.candles import Candles

candles: Candles = all_candles["1h"][0]

# candles.plot(type="returns")
candles.plot(type="candles" )
```

```python
df = candles.data
df['returns'] = df['close'] - df['open']
print(df)
print("Mean return:", df['returns'])
```
